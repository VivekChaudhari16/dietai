```python
# e).1 Outlier Detection:(IQR Method)
def detect_outliers_by_IQR(df, factor):
    numerical_cols = df.select_dtypes(include=np.number).columns
    result = []
    for col in numerical_cols:
        Q1 = np.quantile(df[col], 0.25)
        Q3 = np.quantile(df[col], 0.75)
        IQR = Q3 - Q1
        lower_limit = Q1 - (factor * IQR)
        upper_limit = Q3 + (factor * IQR)
        outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]
        result.append({
                       'Column': col,
                       'Lower Limit': lower_limit,
                       'Upper Limit': upper_limit,
                       'No. of Outliers': outliers.shape[0]
                        })
    return pd.DataFrame(result)
detect_outliers_by_IQR(df,1.5)
```

```python
# KS Test

h0='Age column follows the normal distribution'
h1='Age column does NOT follow normal distribution'

print(f'h0 : {h0}')
print(f'h1: {h1}\n')

x=df['Age']
z_val=(x - x.mean()) / x.std()
Statistic,pvalue=stats.kstest(z_val,'norm')
print(f'Statistic :{Statistic}\npvalue :{pvalue}\n')

if pvalue>0.05:
    print('Fail to reject H0 (Age distribution fits normal distribution)')
else:
    print('Reject H0 (Age distribution does NOT fit normal distribution)')
```

```python
# AD test

h0='Age column follows the normal distribution'
h1='Age column does NOT follow normal distribution'

print(f'h0 : {h0}')
print(f'h1: {h1}\n')

x=df['Age']
Statistic,critical_values,significance_level =stats.anderson(x,'norm')
print(f' Statistic :{Statistic}\n critical_values:{critical_values} \n significance_level  :{significance_level }')

print('\t')

for i,j in zip(critical_values,significance_level):
    if i>Statistic:
        print(f"At {j}% significance level i.e H0 ACCEPTED (Age column Fits normal distribution)")
    else:
        print(f"At {j}% significance level i.e H0 REJECTED (Age column Not normally distributed)")
```

```python
sns.boxplot(df['Age'])
```

```python
sorted_val=df['Age'].sort_values()
cdf_values=np.arange(1,len(sorted_val)+1)/len(sorted_val)

plt.plot(sorted_val,cdf_values)
plt.grid()
plt.title('CDF of age column')
plt.xlabel('age')
plt.ylabel('Cummilative Frequency')
```

```python
#chebishews algorith
import sympy as sp

x=np.mean(df['Age'])
sd=np.std(df['Age'])
k = sp.symbols('k')
range_per=[0.95,0.90,0.85,0.80,0.75,0.5]
per=[]
for i in range_per:
    eq = 1 - 1/k**2 - i
    solution = sp.solve(eq, k)
    per.append(solution[1])
res=[]
for k in per:
    l=x-(k*sd)
    u=x+(k*sd)
    res.append((l,u))
for p,q in zip(range_per,res):
    print(f'{int(p*100)}% of Ages are in between {round(q[0],3)} to {round(q[1],3)}\n')
```

#### CLT on Age Column

```python
sample_means=[]
for i in range(1000):
    sample_values=np.random.choice(df['Age'],size=30)
    sample_means.append(np.mean(sample_values))

print(f'standard error : {np.std(df['Age'])/np.sqrt(1000)}')
sns.distplot(sample_means)
plt.title('Sampling Distribution of Sample Means (Age)')
plt.xlabel('Sample Mean Age ')
```

Points bend above the line so maybe lower cholesterol values are not as low as a normal distribution predicts.                      
Points bend below the line so maybe high cholesterol values are less extreme than expected                

Overall Shape is S-curve. That pattern means our Distribution is NOT normal

```python
# KS Test

h0='Cholesterol Data follows the normal distribution'
h1='Cholesterol Data does NOT follow normal distribution'

print(f'h0 : {h0}')
print(f'h1: {h1}\n')

x=df['Cholesterol']
z_val=(x - x.mean()) / x.std()
Statistic,pvalue=stats.kstest(z_val,'norm')
print(f'Statistic :{Statistic}\npvalue :{pvalue}\n')

if pvalue>0.05:
    print('Fail to reject H0 (Cholesterol distribution fits normal distribution)')
else:
    print('Reject H0 (Cholesterol distribution does NOT fit normal distribution)')
```

```python
# AD test

h0='Cholesterol column follows the normal distribution'
h1='Cholesterol column does NOT follow normal distribution'

print(f'h0 : {h0}')
print(f'h1: {h1}\n')

x=df['Cholesterol']
Statistic,critical_values,significance_level =stats.anderson(x,'norm')
print(f' Statistic :{Statistic}\n critical_values:{critical_values} \n significance_level  :{significance_level }')

print('\t')

for i,j in zip(critical_values,significance_level):
    if i>Statistic:
        print(f"At {j}% significance level i.e H0 ACCEPTED (Cholesterol column fits normal distribution)")
    else:
        print(f"At {j}% significance level i.e H0 REJECTED (Cholesterol column not normally distributed)")
```

```python
#Chebsheys Inequality

x=np.mean(df['Cholesterol'])
sd=np.std(df['Cholesterol'])
k = sp.symbols('k')
range_per=[0.95,0.90,0.85,0.80,0.75,0.5]
per=[]
for i in range_per:
    eq = 1 - 1/k**2 - i
    solution = sp.solve(eq, k)
    per.append(solution[1])
res=[]
for k in per:
    l=x-(k*sd)
    u=x+(k*sd)
    res.append((l,u))
for p,q in zip(range_per,res):
    print(f'{int(p*100)}% of Cholesterol Levels are in between {round(q[0],3)} to {round(q[1],3)}\n')
```
